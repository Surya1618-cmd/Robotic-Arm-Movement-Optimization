```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("../data/training_log.csv")
plt.figure(figsize=(8,4))
plt.plot(df["episode"], df["epsilon"], color="green")
plt.xlabel("Episode")
plt.ylabel("Epsilon")
plt.title("Exploration to Exploitation (Epsilon Decay)")
plt.grid(True)
plt.show()
```

```python
df["rolling_dist"] = df["best_distance"].rolling(window=50).mean()

plt.figure(figsize=(8,4))
plt.plot(df["episode"], df["rolling_dist"], label="Rolling Avg Distance (50)")
plt.xlabel("Episode")
plt.ylabel("Distance")
plt.title("Smoothed Distance Trend")
plt.grid(True)
plt.legend()
plt.show()
```
