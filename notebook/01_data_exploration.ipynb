```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../data/training_log.csv")
df.head()
```

```python
plt.figure(figsize=(8,4))
plt.plot(df["episode"], df["total_reward"], label="total Reward")
plt.xlabel("Episode")
plt.ylabel("Reward")
plt.title("Reward vs Episode")
plt.grid(True)
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(8,4))
plt.plot(df["episode"], df["best_distance"], color="orange", label="Best Distance")
plt.xlabel("Episode")
plt.ylabel("Distance")
plt.title("Best Distance to Target Over Training")
plt.grid(True)
plt.legend()
plt.show()
```
